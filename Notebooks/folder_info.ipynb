```python
import pandas as pd
import os

# List your CSV files
csv_files = [
    "Games.csv",
    "Players.csv",
    "LeagueSchedule24_25.csv",
    "LeagueSchedule25_26.csv",
    "PlayerStatistics.csv",
    "PlayerStatisticsExtended.csv",
    "nba_dailyleaders_full_24_25.csv",
    "nba_dailyleaders_full_2526.csv",
    "NBAPlayersStatistics23:24.csv",
    "TeamHistories.csv",
    "TeamStatistics.csv",
    "TeamStatisticsExtended.csv"
]

# Loop through each file
for file in csv_files:
    try:
        df = pd.read_csv(file)
        print(f"\n--- {file} ---")
        print("First row:")
        print(df.iloc[0].to_dict())
        print("Row count:", len(df))
        print("Unique columns:", list(df.columns))
    except Exception as e:
        print(f"Error reading {file}: {e}")
```

## Helper Function

```python
# Helper function: format columns with newline every 5
def format_columns(columns):
    return "\n".join(
        [" , ".join(columns[i:i+5]) for i in range(0, len(columns), 5)]
    )
```

```python
files = [
    "PlayerStatisticsExtended.csv",
    "PlayerStatistics.csv",
    "TeamStatistics.csv",
    "TeamStatisticsExtended.csv",
    "Players.csv"
]

for file in files:
    path = os.path.join(os.getcwd(), file)

    df = pd.read_csv(path, low_memory=False)

    print("\n" + "=" * 100)
    print(file)
    print("=" * 100)

    print("Rows:", len(df))
    print("Columns:", len(df.columns))

    # Look for anything season/year related
    season_cols = [
        c for c in df.columns
        if any(x in c.lower() for x in ["season", "year"])
    ]

    print("\nSeason/year columns:", season_cols)

    for col in season_cols:
        print(f"\n{col}:")
        print("  Unique:", df[col].nunique(dropna=True))
        print("  Values:", sorted(df[col].dropna().astype(str).unique())[:100])

    # Look for dates
    date_cols = [
        c for c in df.columns
        if any(x in c.lower() for x in ["date", "datetime", "timestamp"])
    ]

    print("\nDate columns:", date_cols)

    for col in date_cols:
        dates = pd.to_datetime(df[col], errors="coerce")

        if dates.notna().any():
            print(
                f"  {col}: "
                f"{dates.min().date()} → {dates.max().date()}"
            )

    print("\nFirst row:")
    print(df.iloc[0].to_dict())
```
